# Introducción a Jupyter Notebook

En este material conoceremos **Jupyter Notebook**, una herramienta que nos permite escribir y ejecutar código Python, documentar lo que hacemos y visualizar los resultados en un mismo lugar.

La idea es comprender cómo funciona esta herramienta y por qué puede ser útil cuando comenzamos a aprender programación.

## 1. ¿Qué es Jupyter Notebook?

Jupyter Notebook es una herramienta interactiva que permite escribir y ejecutar código, agregar explicaciones y visualizar resultados dentro de un mismo documento.

A diferencia de un archivo `.py`, donde normalmente escribimos el código en un archivo de Python, un notebook nos permite dividir el contenido en pequeñas secciones llamadas **celdas**.

Esto resulta especialmente útil para aprender, experimentar con código y explicar conceptos de programación paso a paso.

Los notebooks se guardan normalmente con la extensión `.ipynb`.

## 2. ¿Cómo funciona?

Un notebook está compuesto por celdas. Cada una puede cumplir una función diferente.

### 📝 Celdas Markdown

Permiten escribir texto, títulos, listas, explicaciones y fórmulas. Se utilizan para documentar el material y explicar los conceptos.

### 💻 Celdas de código

Permiten escribir instrucciones de Python y ejecutarlas para comprobar qué hacen.

### 📊 Resultados

Cuando ejecutamos una celda de código, el resultado puede aparecer justo debajo de ella. Dependiendo del código, podemos obtener texto, números, gráficos u otros resultados.

Esta combinación permite alternar entre explicaciones y ejemplos prácticos.

```python
# Esto es un bloque o celda de codigo
print("¡Hola, mundo!")
print(1+3)
```

## 3. ¿Cómo ejecutar una celda?

Para ejecutar una celda de código:

1. Haz clic sobre la celda.
2. Presiona `Shift + Enter` (Mayús + Enter) o en la flecha que parece al lado(➤).
3. Observa el resultado que aparece debajo.

También puedes utilizar el botón **Run** de la barra de herramientas.

**Importante:** si una celda depende de una variable creada en otra, debes ejecutar primero la celda que define esa variable.

## 4. Tipos de celdas

En Jupyter Notebook encontrarás principalmente dos tipos de celdas:

- **Code:** para escribir y ejecutar código.
- **Markdown:** para escribir texto y dar formato a las explicaciones.

## 5. Para recordar

- Jupyter Notebook permite combinar texto, código y resultados.
- Los documentos se guardan normalmente en archivos `.ipynb`.
- Las celdas de código permiten ejecutar Python por partes.
- Las celdas Markdown sirven para explicar y organizar el contenido.
